```python
# Rating Distribution
rating_counts = df_feedback['rating'].value_counts().sort_index()
print(rating_counts)
```

```python
# Low Rating Orders
low_ratings = df_feedback[df_feedback['rating'] <= 2]

print("Low Rating reasons")
print(low_ratings['feedback_category'].value_counts())

print("Low Rating Sentiment Breakdown")
print(low_ratings['sentiment'].value_counts())
```

Delivery Time & Delay Analysis

```python
# Calculate Delivery Delay in Minutes
df_orders['delay_minutes'] = (df_orders['actual_delivery_time'] - df_orders['promised_delivery_time']).dt.total_seconds()/60
```

```python
df_orders['is_delayed'] = df_orders['delay_minutes'] > 0
```

```python
print("Delivey Performance Summary")
print("Total Orders:", len(df_orders))
print("On-Time Orders:", (df_orders['delay_minutes'] <=0).sum())
print("Delayed Orders:", df_orders['is_delayed'].sum())
print("Average Delay for late Orders (Mins):", round(df_orders[df_orders['is_delayed']]['delay_minutes'].mean(), 2))
```

```python
# Merge Orders with Feedback to check Rating Impact
df_order_feedback = pd.merge(df_orders, df_feedback, on='order_id', how='inner')
```

```python
print("Average Rating: ON-TIME vs DELAYED ORDERS")
print(df_order_feedback.groupby('is_delayed')['rating'].mean())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
marketing_numeric = df_marketing[['spend', 'revenue_generated', 'clicks', 'impressions', 'conversions', 'roas']].replace([np.inf, -np.inf], np.nan).dropna()

# Calculate Correlation Matrix
corr_matrix = marketing_numeric.corr()
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='Blues', fmt='.2f', linewidths=0.5)

plt.title('Marketing Metrics Correlation Heatmap', fontsize=14)
plt.show()
```

```python
# Combining sales volume and average rating at the product level.
rating_volume_df = pd.merge(
    df_order_items.groupby('product_id')['quantity'].sum().rename('total_units_sold'),
    pd.merge(df_order_items, df_feedback[['order_id', 'rating']], on='order_id').groupby('product_id')['rating'].mean(),
    on='product_id')
```

```python
plt.figure(figsize=(6, 4))
sns.heatmap(rating_volume_df.corr(), annot=True, cmap='Blues', fmt='.2f', linewidths=0.5)

plt.title('Rating vs Order Volume Correlation', fontsize=12)
plt.show()
```

```python
# Merge order items with products to get category-level revenue
category_sales = pd.merge(df_order_items, df_products, on='product_id')
category_sales['total_revenue'] = category_sales['quantity'] * category_sales['price']

# Group by category
top_categories = category_sales.groupby('category')['total_revenue'].sum().reset_index()
top_categories = top_categories.sort_values(by='total_revenue', ascending=False)
```

```python
# Plot Bar Chart
plt.figure(figsize=(10, 5))
sns.barplot(data=top_categories.head(10), x='total_revenue', y='category', palette='Blues_r')

plt.title('Top 10 Product Categories by Revenue', fontsize=14)
plt.xlabel('Total Revenue (₹)')
plt.ylabel('Category')
plt.show()
```

```python
# Convert date column and aggregate daily revenue
df_marketing['date'] = pd.to_datetime(df_marketing['date'])
daily_revenue = df_marketing.groupby('date')['revenue_generated'].sum().reset_index()

# Plot Line Chart
plt.figure(figsize=(12, 5))
sns.lineplot(data=daily_revenue, x='date', y='revenue_generated', color='#1f77b4', linewidth=2)

plt.title('Campaign Revenue Trends Over Time', fontsize=14)
plt.xlabel('Date')
plt.ylabel('Total Revenue Generated (₹)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

Insight - Revenue Consistency: Tracks daily campaign revenue trends over time to identify seasonal spikes, promotional impacts, and revenue dips.

```python
# Plot Scatter Plot
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df_marketing, 
    x='spend', 
    y='conversions', 
    hue='channel', 
    palette='Set2', 
    s=70, 
    alpha=0.8
)

plt.title('Ad Spend vs Conversions (Efficiency Analysis)', fontsize=14)
plt.xlabel('Ad Spend (₹)')
plt.ylabel('Conversions (Orders)')
plt.legend(title='Marketing Channel', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.4)
plt.show()
```

Insight :- Conversion Efficiency: Analyzes how effectively ad spend translates into customer conversions across different marketing channels.
